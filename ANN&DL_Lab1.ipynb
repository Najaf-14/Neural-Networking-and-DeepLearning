```python
# Array methods
import numpy as np

zero_array = np.zeros((2,3))
print("Zero Array")
print(zero_array)

ones_array = np.ones((2,2))
print("Ones Array")
print(ones_array)

full_array = np.full((4,2), 8)
print("Full Array")
print(full_array)

identity_arr = np.eye(3)
print("Identity Array")
print(identity_arr)

print("Empty Array")
print(np.empty(2))

#evenly spaced array
print(np.arange(1,9,1)) # start=1, end<9, increment of 1
print(np.arange(1,9,2))

#specific no of values equally (spaced) between range
print(np.linspace(1,10,4)) # gap between each num is same

#random values(float)
# r_arr = np.random.rand(2,3)
# print(r_arr)

#int
r_arr = np.random.randint(1,10,(2,3))
print(r_arr)
```
